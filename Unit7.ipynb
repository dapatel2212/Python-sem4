```python
import pandas as pd
import bs4
```

```python
soup = bs4.BeautifulSoup(open('imdb.html').read())
print(soup.prettify())
```

```python
movie_blocks = soup.findAll('div',{'class':'lister-item-content'})
print(movie_blocks[0].prettify())
```

```python
m_name = movie_blocks[0].find('a').text
print(m_name)
```

```python
m_name = movie_blocks[0].find('a').get_text()
print(m_name)
```

```python
m_year = movie_blocks[0].find('span',{'class':'lister-item-year'}).text[1:-1]
print(m_year)
```

```python
m_year = movie_blocks[0].find('div',{'class':'inline-block ratings-imdb-rating'}).text.strip()
print(m_year)
```

```python
m_year = movie_blocks[0].find('div',{'class':'inline-block ratings-imdb-rating'}).get('data-value')
print(m_year)
```

```python
Movie_names = []
Movie_years = []
Movie_ratings = []
for i in movie_blocks:
    Movie_names.append(i.find('a').text)
    Movie_years.append(i.find('span',{'class':'lister-item-year'}).text[1:-1])
    Movie_ratings.append(i.find('div',{'class':'inline-block ratings-imdb-rating'}).get('data-value'))
print(Movie_names)
print(Movie_years)
print(Movie_ratings)
```

```python
df = pd.DataFrame({
    'Movie_names':Movie_names,
    'Movie_years':Movie_years,
    'Movie_ratings':Movie_ratings
})
df
```

## SubsLikeScript

```python
import bs4
import pandas as pd
import requests
```

```python
url = "https://subslikescript.com/movie/Titanic-120338"
result = requests.get(url)
print(result)
```

```python
content = result.text
print(content)
```

```python
soup = bs4.BeautifulSoup(content, 'html.parser')
print(soup.prettify())
```

```python
movie = soup.find('h1').text
print(movie)
```

```python
script = soup.findAll('div',{'class' : 'full-script'})
print(script)
```

```python
script1 = []
for i in script:
    print(i.text)
    script1.append(i.text)
```

```python
file = open("Titanic script.txt", "w", encoding="utf-8")
file.write(movie)
file.writelines(script1)
file.close()
```
