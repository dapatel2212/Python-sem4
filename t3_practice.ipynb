##  Pb375 Write a program for web scrapping using BeautifulSoup to scrape the following details from the given link and make a data frame using that scraped data from the page in a given link. You will find 30 news articles with fact checks on this page. You need to scrape the following details from all the articles andstore that in a data frame. Statement of News, Date of News, Source of News.
Link: https://www.politifact.com/factchecks

```python
response = requests.get("https://www.politifact.com/factchecks")
print(response)
```

```python
data = response.text
soup = bs4.BeautifulSoup(data, "html.parser")
print(soup.prettify())
```

```python
script = soup.find("li", {'class': "o-listicle__item"})
print(script)
```

```python
print("News headline : ",script.find("a").text)
print("Date of news : ",script.find("footer").text.split("•")[1])
print("Source of news : ",script.find("a")['href'])
```

```python
script = soup.findAll("li", {'class': "o-listicle__item"})
print(len(script))
```

```python
list_of_news = []
for i in script:
    df = {
        "News headline" : i.find("a").text.strip(),
        "Date of news" : i.find("footer").text.split("•")[1].strip(),
        "Source of news" : i.find("a")['href']
    }
    list_of_news.append(df)
    print(df)

print(len(list_of_news))
```

```python
import pandas as pd
df = pd.DataFrame(list_of_news)
df
```

## Pb376 Write a program of web scrapping using BeautifulSoup to scrape the given data from the following link.On the above link, you’ll find 30 Popular movies. Scrape the Movie Title and Rating of that particular movie and make a Dataframe of the same.
https://editorial.rottentomatoes.com/guide/popular-movies/


```python
response = requests.get("https://editorial.rottentomatoes.com/guide/popular-movies/")
print(response)
```

```python
data = response.text
soup = bs4.BeautifulSoup(data, "html.parser")
print(soup.prettify())
```

```python
script = soup.findAll("div",class_= "block-countdown rkv-block")
print(len(script))
```

```python
list_of_movies = []
for s in script:
    df = {
        "Movie name" : s.find("a", class_="meta-title").text.strip(),
        "Rating" : s.find("span",class_="tMeterScore").text.strip()
    }
    list_of_movies.append(df)

df = pd.DataFrame(list_of_movies)
df
```
